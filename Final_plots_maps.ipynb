##Interactive Spider Plot to visualize used vehicle data scraped in this study.

```python
df = pd.read_csv('combined_data_final_v3.csv')

# Initialize Dash App
app = dash.Dash(__name__)

# Define allowed fuel types, states, and combined states option
allowed_fuel_types = ['Gas', 'Electric', 'Hybrid', 'Plug-In Hybrid', 'Diesel', 'Hydrogen']
allowed_states = ['Florida', 'California', 'Texas', 'Oregon', 'Massachusetts', 'Georgia']
combined_state_option = 'All States'

# Add a new column in the dataframe that combines the selected states
df['combined_states'] = df['state'].apply(lambda x: x if x in allowed_states else combined_state_option)

# Layout
app.layout = html.Div([
    html.H1("Dynamic Spider Plot: Car Marketplace Comparison"),

    html.Label('Select Fuel Type:'),
    dcc.Dropdown(
        id='fuel_type_dropdown',
        options=[{'label': fuel, 'value': fuel} for fuel in allowed_fuel_types if fuel in df['fuelType'].unique()],
        value=allowed_fuel_types[0],  # Default to the first allowed fuel type
        style={'width': '50%'}
    ),

    html.Label('Select State:'),
    dcc.Dropdown(
        id='state_dropdown',
        options=[
            {'label': state, 'value': state} for state in allowed_states if state in df['state'].unique()
        ] + [{'label': 'All States', 'value': combined_state_option}],  # Add 'All States' option
        value=combined_state_option,  # Default to 'All States'
        style={'width': '50%'}
    ),

    html.Label('Select Year Range:'),
    dcc.RangeSlider(
        id='year_slider',
        min=df['year'].min(),
        max=df['year'].max(),
        step=1,
        marks={year: str(year) for year in range(df['year'].min(), df['year'].max()+1, 1)},
        value=[2010, df['year'].max()]  # Default to 2010 as the starting year
    ),

    dcc.Graph(id='spider_plot')
])

# Callback to update the spider plot
@app.callback(
    Output('spider_plot', 'figure'),
    [
        Input('fuel_type_dropdown', 'value'),
        Input('state_dropdown', 'value'),
        Input('year_slider', 'value')
    ]
)
def update_spider_plot(fuel_type, state, year_range):
    # Filter data based on inputs
    if state == combined_state_option:
        # For "All States", include all the allowed states
        filtered_df = df[
            (df['fuelType'] == fuel_type) &
            (df['state'].isin(allowed_states)) &
            (df['year'] >= year_range[0]) &
            (df['year'] <= year_range[1])
        ]
    else:
        # If a specific state is selected
        filtered_df = df[
            (df['fuelType'] == fuel_type) &
            (df['state'] == state) &
            (df['year'] >= year_range[0]) &
            (df['year'] <= year_range[1])
        ]

    # Calculate average values for price_total for each body style and source
    grouped_data = filtered_df.groupby(['bodyStyleCategory', 'source']).agg(
        avg_price=('price_total', 'mean')
    ).reset_index()

    # Define the body styles as labels
    body_styles = df['bodyStyleCategory'].unique()
    labels = [f"{style}" for style in body_styles]  # Keep clean labels

    # Create data for TrueCar (TC) and Carvana (CV)
    tc_data = []
    cv_data = []

    for body_style in body_styles:
        # Get the average price for TC and CV for each body style
        tc_price = grouped_data[(grouped_data['bodyStyleCategory'] == body_style) & (grouped_data['source'] == 'TC')]
        cv_price = grouped_data[(grouped_data['bodyStyleCategory'] == body_style) & (grouped_data['source'] == 'CV')]

        tc_data.append(tc_price['avg_price'].values[0] if len(tc_price) > 0 else 0)
        cv_data.append(cv_price['avg_price'].values[0] if len(cv_price) > 0 else 0)

    # Create the spider plot (Radar Chart)
    fig = go.Figure()

    fig.add_trace(go.Scatterpolar(
        r=tc_data,
        theta=labels,
        fill='toself',
        name='TrueCar (TC)',
        line=dict(color='blue'),
        hovertemplate='%{theta} - Avg. Price: $%{r:,.0f}<extra></extra>'  # Custom hover text
    ))

    fig.add_trace(go.Scatterpolar(
        r=cv_data,
        theta=labels,
        fill='toself',
        name='Carvana (CV)',
        line=dict(color='red'),
        hovertemplate='%{theta} - Avg. Price: $%{r:,.0f}<extra></extra>'  # Custom hover text
    ))

    fig.update_layout(
        polar=dict(
            radialaxis=dict(

                visible=True,
                range=[0, max(max(tc_data), max(cv_data)) + 10000]
            )
        ),
        title=f"Avg. Price Comparison: {fuel_type} ({state}) - TrueCar vs. Carvana",
        template="plotly_dark"
    )

    return fig

# Run app
app.run_server(mode='inline')
```

##Interactive Maps

```python

df = pd.read_csv('combined_data_final_v3.csv')

df = df.dropna(subset=['latitude', 'longitude'])
df = df.dropna(subset=['fuelType'])
df = df.dropna(subset=['place'])

df['fuelType'] = df['fuelType'].str.lower()

class_counts = df['fuelType'].value_counts()

sampled_df_list = []

for fuel_type, count in class_counts.items():
    category_df = df[df['fuelType'] == fuel_type]

    if count < 200:
        sampled_df_list.append(category_df)
    elif count > 1000:
        sampled_df_list.append(category_df.sample(1000, random_state=42))
    else:
        sample_count = int(count * 0.2)
        sampled_df_list.append(category_df.sample(frac=0.2, random_state=42))

sampled_df = pd.concat(sampled_df_list)

if len(sampled_df) > 2500:
    sampled_df = sampled_df.sample(2500, random_state=42)

# print("\nSampled Categories and Counts:")
# print(sampled_df['fuelType'].value_counts())

df = sampled_df.dropna(subset=['price_total'])

df['price_total'] = df['price_total'].replace({'\$': '', ',': ''}, regex=True)
df['price_total'] = pd.to_numeric(df['price_total'], errors='coerce')

df = df.dropna(subset=['price_total'])
df = df[df['price_total'] > 0]

price_bins = [0, 10000, 20000, 30000, 40000, 50000, float('inf')]
price_labels = ['$0-$10k', '$10k-$20k', '$20k-$30k', '$30k-$40k', '$40k-$50k', 'Above $50k']

df['Price Range'] = pd.cut(df['price_total'], bins=price_bins, labels=price_labels, right=False)
new_sampled_df = df

m = folium.Map(location=[39.8283, -98.5795], zoom_start=4)

price_range_colors = {
    '$0-$10k': 'blue',
    '$10k-$20k': 'green',
    '$20k-$30k': 'orange',
    '$30k-$40k': 'red',
    '$40k-$50k': 'purple',
    'Above $50k': 'black'
}

feature_groups = {}

for price_range in price_labels:
    feature_groups[price_range] = FeatureGroup(name=price_range)

for _, row in new_sampled_df.iterrows():
    price_range = row['Price Range']
    location = [row['latitude'], row['longitude']]
    popup_content = f"Location: {row['place']}\n\nPrice: ${row['price_total']:<,.2f}"

    marker_color = price_range_colors.get(price_range, 'gray')

    folium.Marker(
        location=location,
        popup=folium.Popup(popup_content, parse_html=True),
        tooltip=row['place'],
        icon=folium.Icon(color=marker_color)
    ).add_to(feature_groups[price_range])

for fg in feature_groups.values():
    fg.add_to(m)

LayerControl(collapsed=False).add_to(m)

display(m)

m.save('sampled_categories_cars.html')
# print("Map successfully exported to 'sampled_categories_cars.html'")
```

```python
df = pd.read_csv('combined_data_final_v3.csv', dtype={'latitude': float, 'longitude': float, 'fuelType': str, 'price_total': float}, low_memory=False)

df = df.dropna(subset=['latitude', 'longitude', 'fuelType', 'place', 'state'])

df.loc[:, 'fuelType'] = df['fuelType'].str.lower()

def categorize_state(state):
    state_keywords = {
        'california': 'California',
        'georgia': 'Georgia',
        'florida': 'Florida',
        'texas': 'Texas',
        'massachusetts': 'Massachusetts',
        'oregon': 'Oregon'
    }

    state_lower = state.lower()

    for state_name, full_state in state_keywords.items():
        if state_name in state_lower:
            return full_state

    return 'Other States'

df.loc[:, 'state_category'] = df['state'].apply(categorize_state)

top_10_percent = df.groupby('fuelType').apply(
    lambda x: x.nsmallest(int(len(x) * 0.1), 'price_total')
).reset_index(drop=True)

gas_top_10_percent = top_10_percent[top_10_percent['fuelType'] == 'gas'].head(1000)

top_10_percent_final = pd.concat([
    top_10_percent[top_10_percent['fuelType'] != 'gas'],
    gas_top_10_percent
]).reset_index(drop=True)

m = folium.Map(location=[39.8283, -98.5795], zoom_start=4, control_scale=True)

fuel_type_colors = {
    'gas': 'blue',
    'diesel': 'red',
    'electric': 'green',
    'hybrid': 'orange',
    'hydrogen': 'purple',
    'plug-in hybrid': 'yellow',
    'cng': 'black'
}

state_layers = {}
fuel_type_layers = {}

for _, row in top_10_percent_final.iterrows():
    fuel_type = row['fuelType']
    location = [row['latitude'], row['longitude']]

    popup_content = f"{row['place']}\n ${int(row['price_total'])}"

    state_category = row['state_category']
    if state_category not in state_layers:
        state_layers[state_category] = FeatureGroup(name=state_category)

    if fuel_type not in fuel_type_layers:
        fuel_type_layers[fuel_type] = FeatureGroup(name=fuel_type.capitalize())

    marker_color = fuel_type_colors.get(fuel_type, 'gray')

    folium.Marker(
        location=location,
        popup=folium.Popup(popup_content, parse_html=True),
        tooltip=row['place'],
        icon=folium.Icon(color=marker_color)
    ).add_to(state_layers[state_category])

    folium.Marker(
        location=location,
        popup=folium.Popup(popup_content, parse_html=True),
        tooltip=row['place'],
        icon=folium.Icon(color=marker_color)
    ).add_to(fuel_type_layers[fuel_type])

state_order = ['California', 'Georgia', 'Florida', 'Texas', 'Massachusetts', 'Oregon', 'Other States']
for state in state_order:
    if state in state_layers:
        state_layers[state].add_to(m)

for layer in fuel_type_layers.values():
    layer.add_to(m)

LayerControl(collapsed=False).add_to(m)

title_html = '''
             <h3 align="center" style="font-size:20px"><b>Top 10% Cheapest Cars by Fuel Type </b></h3>
             '''
m.get_root().html.add_child(folium.Element(title_html))

display(m)

m.save('top_10_percent_cheapest_cars.html')

# print("Map successfully exported to 'top_10_percent_cheapest_cars.html'")
```

```python
df = pd.read_csv('combined_data_final_v3.csv')

df = df.dropna(subset=['latitude', 'longitude'])
df = df.dropna(subset=['fuelType'])
df = df.dropna(subset=['place'])

df['fuelType'] = df['fuelType'].str.lower()

def categorize_state(state):

    state_keywords = {
        'california': 'California',
        'georgia': 'Georgia',
        'florida': 'Florida',
        'texas': 'Texas',
        'massachusetts': 'Massachusetts',
        'oregon': 'Oregon'
    }

    state_lower = state.lower()

    for state_name, full_state in state_keywords.items():
        if state_name in state_lower:
            return full_state

    return 'Other States'

df.loc[:, 'state_category'] = df['state'].apply(categorize_state)

class_counts = df['fuelType'].value_counts()

sampled_df_list = []

for fuel_type, count in class_counts.items():
    category_df = df[df['fuelType'] == fuel_type]

    if count < 200:
        sampled_df_list.append(category_df)
    elif count > 1000:
        sampled_df_list.append(category_df.sample(1000, random_state=42))
    else:
        sample_count = int(count * 0.2)
        sampled_df_list.append(category_df.sample(frac=0.2, random_state=42))

sampled_df = pd.concat(sampled_df_list)

sampled_df['Price'] = sampled_df['price_total'].replace({'\$': '', ',': ''}, regex=True)
sampled_df['Price'] = pd.to_numeric(sampled_df['Price'], errors='coerce')

sampled_df = sampled_df.dropna(subset=['Price'])
sampled_df = sampled_df[sampled_df['Price'] > 0]

price_bins = [0, 10000, 20000, 30000, 40000, 50000, float('inf')]
price_labels = ['$0-$10k', '$10k-$20k', '$20k-$30k', '$30k-$40k', '$40k-$50k', 'Above $50k']
sampled_df['Price Range'] = pd.cut(sampled_df['Price'], bins=price_bins, labels=price_labels, right=False)

fuel_types = sampled_df['fuelType'].dropna().unique()

states = sampled_df['state_category'].dropna().unique()

state_order = ['California', 'Georgia', 'Florida', 'Texas', 'Massachusetts', 'Oregon', 'Other States']

states_sorted = [state for state in state_order if state in states]

def update_map(fuel_types_selected, price_ranges, states_selected):
    m = folium.Map(location=[39.8283, -98.5795], zoom_start=4)

    price_range_colors = {
        '$0-$10k': 'blue',
        '$10k-$20k': 'green',
        '$20k-$30k': 'orange',
        '$30k-$40k': 'red',
        '$40k-$50k': 'purple',
        'Above $50k': 'black'
    }

    state_layers = {state: FeatureGroup(name=state) for state in states_selected}

    filtered_df = sampled_df[sampled_df['fuelType'].isin(fuel_types_selected)]
    filtered_df = filtered_df[filtered_df['Price Range'].isin(price_ranges)]
    filtered_df = filtered_df[filtered_df['state_category'].isin(states_selected)]

    for _, row in filtered_df.iterrows():
        price_range = row['Price Range']
        location = [row['latitude'], row['longitude']]
        popup_content = f"{row['place']}\n${row['Price']:<,.2f}\n{row['fuelType']}\n{row['state_category']}"

        marker_color = price_range_colors.get(price_range, 'gray')

        state_layers[row['state_category']].add_child(
            folium.Marker(
                location=location,
                popup=folium.Popup(popup_content, parse_html=True),
                tooltip=row['place'],
                icon=folium.Icon(color=marker_color)
            )
        )

    for state in state_order:
        if state in state_layers:
            state_layers[state].add_to(m)

    LayerControl(collapsed=False).add_to(m)

    display(m)

fuel_type_selector = widgets.SelectMultiple(
    options=fuel_types.tolist(),
    value=fuel_types.tolist(),
    description='Fuel Type:',
    disabled=False
)

price_range_selector = widgets.SelectMultiple(
    options=price_labels,
    value=price_labels,
    description='Price Range:',
    disabled=False
)

state_selector = widgets.SelectMultiple(
    options=states_sorted,
    value=states_sorted,
    description='States:',
    disabled=False
)

ui = widgets.HBox([fuel_type_selector, price_range_selector, state_selector])

out = widgets.interactive_output(update_map, {
    'fuel_types_selected': fuel_type_selector,
    'price_ranges': price_range_selector,
    'states_selected': state_selector
})

display(ui, out)
```
